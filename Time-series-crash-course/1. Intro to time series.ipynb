
# Ví dụ Biểu đồ Time Series


```python
# Import các package
import yfinance as yf
import plotly.graph_objects as go
```

```python
# Định nghĩa mã cổ phiếu và khoảng thời gian
ticker_symbol = 'AAPL'
start_date = '2021-01-01'
end_date = '2023-01-01'
```

```python
# Lấy dữ liệu từ Yahoo Finance
data = yf.download(ticker_symbol, start=start_date, end=end_date)
```

```python
# Xem dữ liệu
data
```

```python
# Tạo biểu đồ
fig = go.Figure()
fig.add_trace(go.Scatter(x=data.index, y=data['Close'], mode='lines', name='Giá đóng cửa'))
fig.update_layout(title=f'Giá cổ phiếu {ticker_symbol} từ Yahoo Finance',
                  xaxis_title='Ngày',
                  yaxis_title='Giá',
                  template='simple_white',
                  width=900,
                  height=500) 
fig.show()
```
